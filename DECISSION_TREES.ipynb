# :REGRESSION TREES

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from pandas_datareader import data
from sklearn import metrics
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import GridSearchCV
```

```python
boston = fetch_california_housing()
df = pd.DataFrame(boston.data)
```

```python
df.columns = boston.feature_names
df['target'] = boston.target
```

```python
df.head()
```

```python
x = df.iloc[: , 0:8]
y = df.iloc[: , 8]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
dt = DecisionTreeRegressor( criterion = 'squared_error', max_depth=5)
dt.fit(x_train,y_train)
```

```python
ypred = dt.predict(x_test)
accuracy = metrics.r2_score(y_test,ypred)
print(accuracy)
```

```python
param_grid = {
    'max_depth':[2,4,5,None],
    'splitter' : ['best','random'],
    'min_samples_leaf' : [4,2,5],
    'min_samples_split' : [2,4,5,6,7],
    'min_impurity_decrease' : [0.001,0.01],
    'max_features':[None, 4,2,'sqrt']

}
```

```python
grid_search = GridSearchCV (estimator=dt,param_grid=param_grid,cv=5)
grid_search.fit(x_train,y_train)
```

```python
grid_search.best_score_
```

```python
grid_search.best_params_
```
